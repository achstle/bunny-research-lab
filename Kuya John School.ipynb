```python
import ipywidgets as widgets
from IPython.display import display

carrot_count = 0

label = widgets.Label(value=f"Carrots Eaten: {carrot_count}")

button = widgets.Button(
    description='Feed a Carrot! 🥕',
    style={
        'button_color': '#FFD1DC',
        'text_color': '#D10056',
        'font_weight': 'bold'
    }
)

def on_button_clicked(b):
    global carrot_count
    carrot_count += 1
    label.value = f"Carrots Eaten: {carrot_count}"

button.on_click(on_button_clicked)

display(label, button)
```
